# Grafik Peningkatan Performa Recommender System

Notebook ini memvisualisasikan peningkatan performa model recommender dari eksperimen Baseline, ke Lightweight Fine-Tuning, dan terakhir ke Partial Unfreeze Fine-Tuning.

```python

import matplotlib.pyplot as plt
import numpy as np

# Data
phases = ['Baseline', 'Lightweight', 'Partial Unfreeze']

# Data for VGG19 (Main Model)
vgg19_p5 = [0.2700, 0.3900, 0.6780]
vgg19_ndcg5 = [0.2672, 0.3954, 0.6742]

# Data for Overall Best Model per Phase
# Baseline: InceptionV3, Lightweight: InceptionV3, Partial Unfreeze: VGG19
best_p5 = [0.3380, 0.5160, 0.6780]
best_ndcg5 = [0.3400, 0.5248, 0.6742]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Improvement of VGG19
ax1.plot(phases, vgg19_p5, marker='o', label='Precision@5', color='#2ca02c', linewidth=2, markersize=8)
ax1.plot(phases, vgg19_ndcg5, marker='s', label='NDCG@5', color='#ff7f0e', linewidth=2, markersize=8, linestyle='--')
ax1.set_title('Peningkatan Performa Recommender: VGG19', fontsize=14)
ax1.set_ylabel('Score', fontsize=12)
ax1.set_ylim(0.2, 0.8)
ax1.grid(True, linestyle='--', alpha=0.7)
ax1.legend(fontsize=12)

# Annotate Plot 1
for i, (p, n) in enumerate(zip(vgg19_p5, vgg19_ndcg5)):
    ax1.annotate(f"{p:.4f}", (phases[i], p), textcoords="offset points", xytext=(0,10), ha='center', fontsize=10, weight='bold')
    ax1.annotate(f"{n:.4f}", (phases[i], n), textcoords="offset points", xytext=(0,-15), ha='center', fontsize=10, weight='bold')

# Plot 2: Improvement of Best Model per Phase
ax2.plot(phases, best_p5, marker='o', label='Precision@5', color='#1f77b4', linewidth=2, markersize=8)
ax2.plot(phases, best_ndcg5, marker='s', label='NDCG@5', color='#d62728', linewidth=2, markersize=8, linestyle='--')
ax2.set_title('Peningkatan Performa Recommender: Model Terbaik Tiap Fase', fontsize=14)
ax2.set_ylabel('Score', fontsize=12)
ax2.set_ylim(0.3, 0.8)
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.legend(fontsize=12)

# Annotate Plot 2
best_model_labels = ['(InceptionV3)', '(InceptionV3)', '(VGG19)']
for i, (p, n, m) in enumerate(zip(best_p5, best_ndcg5, best_model_labels)):
    ax2.annotate(f"{p:.4f}\n{m}", (phases[i], p), textcoords="offset points", xytext=(0,10), ha='center', fontsize=10, weight='bold')
    ax2.annotate(f"{n:.4f}", (phases[i], n), textcoords="offset points", xytext=(0,-15), ha='center', fontsize=10, weight='bold')

plt.tight_layout()
plt.show()
```
